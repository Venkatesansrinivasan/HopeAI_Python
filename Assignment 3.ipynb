```python
# Catgory the people by their age like children,adult,citizen,senior citizen...
age = int(input("age: "))
if age < 13:
    print("Children")
elif age < 20:
    print ("Teenager")
elif age < 60:
    print("Adult")
elif age < 80:
    print("Senior Citizen")
else:
    print("Senior Citizen")
```
