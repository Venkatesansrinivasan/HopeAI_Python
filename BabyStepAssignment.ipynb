```python
#Body mass index (BMI)
bmi = float(input("Enter the input:"))
           
if bmi < 18.5:
    print("Under Weight")
    
elif bmi<124.9:
    print("Normal Weight")
    
elif bmi<29.9:
    print("Over Weight")
    
elif bmi>=30:
    print("obese")
    
elif bmi>=30.0-34.9:
    print("Obese Class1")
    
elif bmi>=35.0-39.9:
    print("Obese Class2")
    
elif bmi>=40:
    print("Obese Class3")
```
